```python
import heapq

def astar(g, h, start, goal):
    q = [(h[start], 0, start, [start])]

    while q:
        f, c, n, p = heapq.heappop(q)

        if n == goal:
            return p, c

        for x, w in g[n]:
            new_cost = c + w
            heapq.heappush(
                q, (new_cost + h[x], new_cost, x, p + [x])
            )

    return None, float("inf")


g = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 2)],
    'C': [('E', 2)],
    'D': [('G', 3)],
    'E': [('G', 1)],
    'G': []
}

h = {
    'A': 6,
    'B': 4,
    'C': 3,
    'D': 2,
    'E': 1,
    'G': 0
}

p, c = astar(g, h, 'A', 'G')

print("Path:", " -> ".join(p) if p else "No path found")
print("Cost:", c)
```
