```python
import math

def alpha_beta(depth, nodeIndex, maximizingPlayer, values, alpha, beta, height):

    if depth == height:
        return values[nodeIndex]

    if maximizingPlayer:
        best = -math.inf
        for i in range(2):
            val = alpha_beta(depth + 1, nodeIndex * 2 + i, False, values, alpha, beta, height)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = math.inf
        for i in range(2):
            val = alpha_beta(depth + 1, nodeIndex * 2 + i, True, values, alpha, beta, height)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

def main():
    while True:
        try:
            values_str = input("Enter leaf node values (comma-separated, e.g., 3,5,2,9): ")
            values = [int(x.strip()) for x in values_str.split(',')]
            n_leaf_nodes = len(values)
            if n_leaf_nodes > 0 and (n_leaf_nodes & (n_leaf_nodes - 1) == 0): # Check if n_leaf_nodes is a power of 2
                break
            else:
                print("Number of leaf nodes must be a power of 2 and greater than 0. Please try again.")
        except ValueError:
            print("Invalid input. Please enter a comma-separated list of integers.")

    print(f"Leaf node values: {values}")

    height = int(math.log2(n_leaf_nodes))

    optimal_value = alpha_beta(0, 0, True, values, -math.inf, math.inf, height)

    print(f"\nOptimal value using Alpha-Beta Pruning: {optimal_value}")

main()
```
